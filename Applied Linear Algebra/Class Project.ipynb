```python
def back_substitution(R,b):
    n = R.shape[0]
    x = np.zeros(n, dtype = float)
    for i in range (n - 1, -1, -1):
        x[i] = (b[i] - np.dot(R[i, i+1:], x[i+1:]))/ R[i,i]
    return x   

A = np.array([[2,1],
              [1,-1],
              [3,2],
              [4,1]])
y = np.array([1, 2, 3, 4], dtype=float)

# QR decomposition (reduced)
Q, R = np.linalg.qr(A, mode='reduced')

# Compute Q^T y
b = Q.T @ y

x_back = back_substitution(R,b)
res_back = np.linalg.norm(A @ x_back - y, 2)

x, residuals, rank, sv = np.linalg.lstsq(A,y)

print("x (lstsq):", x)
print("residuals:", residuals)
print("x_back (QR):", x_back)
print("res_back:", res_back)
```

```python
A = np.array([[4,1,-2],
              [1,2,0],
              [-2,0,3]])

def eigval(A, iterations):
    A_k = A.astype(float).copy()
    m, n = A_k.shape
       
    def decomp(A_i):
        m, n = A_i.shape
        Q = np.zeros((m,n))
        R = np.zeros((n,n))

        for j in range(n):
            v = A_i[:, j].copy()
            for i in range(j):
                R[i,j] = np.dot(Q[:, i], A_i[:,j])
                v = v - R[i,j] * Q[:,i]
            R[j,j] = np.linalg.norm(v) 
            Q[:,j] = v/R[j,j] 
        return Q, R

    for _ in range(iterations):
        Q, R = decomp(A_k)
        A_k = R @ Q

    eigvals = np.diag(A_k)
    return eigvals 


iterations = [1, 5, 8, 10, 100, 1000]

for iter_count in iterations:
    eiggg_ = eigval(A, iter_count)
    print("Our Function at " + str(iter_count) + " iterations")
    print(np.sort(eiggg_))
    print()

eigenvalues = np.linalg.eig(A)

print("Numpy Function")

print(np.sort(eigenvalues.eigenvalues))
```
