```python
# Part A
import numpy as np 
A = np.array([[1,3,3],
            [2,2,-2],
            [-2,2,1]], dtype= float)
m, n = A.shape
Q = np.zeros((m,n))
R = np.zeros((n,n))

for j in range(n):
    v = A[:, j]
    for i in range(j):
        R[i,j] = np.dot(Q[:, i], A[:,j])
        v = v - R[i,j] * Q[:,i]
    R[j,j] = np.linalg.norm(v) 
    Q[:,j] = v/R[j,j] 
np.linalg.inv(Q)
print(Q)
print(R)

# Part B
Q_np, R_np = np.linalg.qr(A)
print(Q_np)
print(R_np)
```
